```python
import math

def euler_criterion(a, p):
    if p == 2:
        return True
    return pow(a, (p - 1) // 2, p) == 1

def is_prime(n):
    if n < 2: return False
    for i in range(2, int(math.isqrt(n)) + 1):
        if n % i == 0: return False
    return True

def get_primes(B):
    return [p for p in range(2, B + 1) if is_prime(p)]

def dixon_algorithm(n, B):
    primes = get_primes(B)
    factor_base = [p for p in primes if euler_criterion(n, p)]
    start = math.ceil(math.isqrt(n))
    relations = []
    x = start

    while len(relations) < len(factor_base) + 2:
        target = (x * x) % n
        temp = target
        factors = []
        for p in factor_base:
            count = 0
            while temp % p == 0:
                count += 1
                temp //= p
            factors.append(count)

        if temp == 1:
            relations.append((x, target, factors))
        x += 1
    #help from geeksforgeeks at this point
    M = []
    for r in relations:
        M.append([val % 2 for val in r[2]])

    num_relations = len(relations)
    num_primes = len(factor_base)

    pivots = []
    matrix = [row[:] for row in M]
    for i in range(num_relations):
        matrix[i] = matrix[i] + [1 if j == i else 0 for j in range(num_relations)]

    col = 0
    for r in range(num_primes):
        while col < num_relations and all(matrix[i][col] == 0 for i in range(r, num_relations)):
            col += 1
        if col >= num_relations:
            break
        pivot_row = next(i for i in range(r, num_relations) if matrix[i][col] == 1)
        matrix[r], matrix[pivot_row] = matrix[pivot_row], matrix[r]
        pivots.append((r, col))

        for i in range(num_relations):
            if i != r and matrix[i][col] == 1:
                matrix[i] = [a ^ b for a, b in zip(matrix[i], matrix[r])]
        col += 1

    for r in range(num_relations):
        if all(val == 0 for val in matrix[r][:num_primes]):
            combination = matrix[r][num_primes:]
            indices = [idx for idx, val in enumerate(combination) if val == 1]
            if not indices:
                continue
       #until here...
            X = 1
            Y_sq = 1
            for idx in indices:
                X *= relations[idx][0]
                Y_sq *= relations[idx][1]

            X = X % n
            Y = math.isqrt(Y_sq) % n

            if X != Y and X != (n - Y):
                d1 = math.gcd(X - Y, n)
                d2 = n // d1
                if 1 < d1 < n:
                    return d1, d2

dixon_algorithm(98069, 47)
```
